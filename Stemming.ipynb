```python
words = [
    "playing",
    "played",
    "plays",
    "player",
    "studies",
    "studying",
    "studied",
    "connected",
    "connecting",
    "connection",
    "connections",
    "running",
    "runs",
    "runner",
    "easily",
    "easier",
    "easiest",
    "happiness",
    "happy",
    "happily"
]
```


**PorterStemmer**

```python
from nltk.stem import PorterStemmer
```

```python
stemming = PorterStemmer()
```

```python
for word in words:
  print(word+"-->"+stemming.stem(word))
```

RegexpStemmer


```python
from nltk.stem import RegexpStemmer

reg_stemmer = RegexpStemmer(r'ing$|s$|ed$|able$', min=3)

print(reg_stemmer.stem('playing'))
```

  SnowBall Stemmer

```python
from nltk.stem import SnowballStemmer
stemmer = SnowballStemmer(language='english')
```

```python
for i in words:
  print(i + "-->" + stemmer.stem(i))
```
